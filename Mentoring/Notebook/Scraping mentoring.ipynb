#Import Library

```python
import pandas as pd
from youtube_comment_downloader import YoutubeCommentDownloader
pd.set_option('display.max_colwidth', None)
```

#Sracping Comments Youtube

```python
downloader = YoutubeCommentDownloader()

# Daftar URL YouTube
urls = [
    "https://www.youtube.com/watch?v=0BRFlBVWH3c",
    "https://www.youtube.com/watch?v=a2WXt0aW76g"
]

all_comments = []

for url in urls:
    print(f"Mengambil komentar dari: {url}")

    count = 0
    try:
        for comment in downloader.get_comments_from_url(url):
            comment["video_url"] = url  # Menambahkan informasi asal video
            all_comments.append(comment)

            count += 1
            if count == 5000:      # Maksimal 5000 komentar per video
                break

    except Exception as e:
        print(f"Gagal mengambil komentar dari {url}")
        print(e)

# Mengubah menjadi DataFrame
df_youtube = pd.DataFrame(all_comments)

df_youtube.head()
```

#Data Cleaning Sementara

##Ubah Tipe Data

```python
#mengubah data pada kolom Time
df_youtube['time'] = pd.to_datetime(
    df_youtube['time_parsed'], unit='s'
).dt.strftime('%d-%m-%Y %H:%M:%S')
```
